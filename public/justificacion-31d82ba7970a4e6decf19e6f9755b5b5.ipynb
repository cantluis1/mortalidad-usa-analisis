# 15. Justificación del Modelo Seleccionado

La selección del modelo óptimo se realiza con base en el **AICc** (criterio de Akaike corregido), métrica apropiada para n < 40, complementada por el diagnóstico de residuos (Ljung-Box) y la coherencia teórica con la prueba ADF.

## Resumen de evidencia

| Criterio | ARIMA | ETS | Holt lineal | Holt amort. | Regresión |
|----------|-------|-----|-------------|-------------|-----------|
| AICc (menor = mejor) | **130.43** | 137.84 | 142.08 | 148.39 | N/A (I(1)) |
| Parámetros | 1 | 4 | 4 | 5 | 2 |
| Ljung-Box p > 0.05 | ✔ | ✔ | — | — | ⚠ |
| Teóricamente correcto para I(1) | ✔ | ✗ | ✗ | ✗ | ✗ |

## Por qué ARIMA(0,1,0) with drift

ARIMA obtiene el AICc más bajo: 130.43, con un único parámetro estimado — el drift (β ≈ 3.24 por año). Es el modelo más parsimonioso, sus residuos son independientes (Ljung-Box: p > 0.05), y es teóricamente correcto para la serie: la diferenciación de orden 1 elimina la raíz unitaria confirmada por ADF, y el drift captura el crecimiento estructural coherente con la aceleración documentada durante la epidemia de opioides desde 2014.

ETS obtiene AICc = 137.84, una diferencia de 7.41 puntos frente a ARIMA. Según Burnham & Anderson (2002), diferencias superiores a 7 puntos en AICc constituyen evidencia fuerte contra el modelo con mayor valor.

La regresión obtiene buenos resultados en tsCV, pero la prueba ADF confirmó que la serie es I(1) (no estacionaria), y aplicar OLS sobre una serie no estacionaria sin diferenciar viola el supuesto de errores independientes. Sus intervalos de predicción son inválidos para la serie I(1) (Granger & Newbold, 1974).

Sus proyecciones indican que la tasa de Unintentional injuries en West Virginia continuará creciendo hacia 2022, profundizando la brecha con el promedio nacional identificada en la sección 9.3 y demandando intervención prioritaria en política de salud pública.
