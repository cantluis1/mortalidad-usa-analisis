# 16. Insights Clave

**El poder obstinado y omnipresente de las enfermedades crónicas:** Las enfermedades cardíacas y el cáncer siguen siendo las dos principales causas de muerte en todo el período (1999–2017). Aunque ambas lideran, las enfermedades cardíacas están experimentando un descenso constante (de 260 a 165 por cada 100,000 personas), lo que refleja mejoras en la prevención y el tratamiento, mientras que el cáncer se mantiene más estable. Este patrón corrobora el dominio de las enfermedades crónicas en el cambio epidemiológico (Shah et al. 2019).

**Crisis silenciosa de las "muertes por desesperación":** Las muertes por suicidio y las lesiones accidentales están aumentando, con estas últimas superando las 48 por cada 100,000 en 2017 y acelerándose desde 2014. Este efecto está fuertemente vinculado a la epidemia de trastornos por uso de opioides y a la pobreza (Daugherty et al. 2019).

**Profundas y obstinadas brechas regionales:** West Virginia, con 957 por cada 100,000 personas, casi dobla a Hawái (585). Este desequilibrio en el sur rural y en los Apalaches tiende a permanecer en el tiempo, reflejando un estatus socioeconómico más bajo y un peor acceso a la atención médica (James et al. 2018).

**El Alzheimer emerge como un desafío creciente:** El Alzheimer sube del lugar 8 al 6, con un aumento de muertes del 173% — el porcentaje más elevado entre todas las causas. Su tendencia ascendente muestra el envejecimiento de la población (Shah et al. 2019).

**El ranking es dinámico, pero el top 2 es inamovible:** El stroke (de la posición 3 a la 5) baja, y las lesiones no intencionales (del puesto 5 al 3) y el Alzheimer (del 8 al 6) suben, pero las afecciones del corazón y el cáncer permanecen sin cambios en los dos primeros puestos.

**La carga de mortalidad se ubica en ciertas zonas geográficas:** El total de muertes es mayor en California, Texas, Florida y Nueva York debido a la alta densidad poblacional. Sin embargo, esto contrasta con índices ajustados bajos, lo que muestra la disparidad entre carga absoluta (demanda de asistencia) y riesgo relativo (probabilidad de fallecer).
