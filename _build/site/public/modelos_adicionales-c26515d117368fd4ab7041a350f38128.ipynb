# 11. Modelos Adicionales

Esta sección presenta los modelos alternativos evaluados para proyectar la tasa de mortalidad por lesiones no intencionales en West Virginia: ETS (Suavización Exponencial) y Regresión Lineal. Ambos se comparan con ARIMA en la sección 14.

Los tres modelos operan sobre la misma serie: tasa ajustada de Unintentional injuries en West Virginia (1999–2017), n = 19 observaciones.
